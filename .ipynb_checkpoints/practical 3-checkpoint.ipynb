# Data Preprocessing: Handling missing values, duplicate removal, encoding categorical variables, normalization, standardization and train-test split

```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
# Load dataset
df = sns.load_dataset('titanic')
```

```python
# Select required columns
df = df[['survived', 'pclass', 'sex', 'age', 'fare', 'embarked']].copy()
```

```python
# Check missing values
print(df.isnull().sum())
```

```python
# Fill missing values
df['age'] = df['age'].fillna(df['age'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
# Remove duplicates
df = df.drop_duplicates()
```

```python
# Encode categorical variables
df = pd.get_dummies(df, columns=['sex', 'embarked'], drop_first=True, dtype=int)
```

```python
# Separate features and target
X = df.drop('survived', axis=1)
y = df['survived']
```

```python
# Normalization
scaler = MinMaxScaler()
X_normalized = scaler.fit_transform(X)
```

```python
# Standardization
scaler2 = StandardScaler()
X_standardized = scaler2.fit_transform(X)
```

```python
# Train test split
X_train, X_test, y_train, y_test = train_test_split(
    X_normalized, y, test_size=0.2, random_state=42
)
```

```python
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Preprocessing completed")
```
