# Introduction to Python ML Environment: Working with NumPy arrays, Pandas DataFrames, importing datasets, dataset inspection, statistical summary functions and dataset information analysis

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
# Load Iris dataset
from sklearn.datasets import load_iris
data = load_iris()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
```

```python
# Display first 5 rows
print(df.head())
```

```python
# Display last 5 rows
print(df.tail())
```

```python
# Dataset information
print(df.info())
```

```python
# Statistical summary
print(df.describe())
```

```python
# Number of rows and columns
print(df.shape)
```

```python
# Column names
print(df.columns)
```

```python
# Missing values
print(df.isnull().sum())
```

```python
# Check duplicate records
print(df.duplicated().sum())
```

```python
# Unique values
print(df.nunique())
```
