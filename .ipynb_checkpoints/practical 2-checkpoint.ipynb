# Data Exploration and Visualization: Scatter plot, line plot, histogram, box plot, heatmap, correlation analysis and outlier visualization

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv('insurance.csv')
```

```python
# Display data
print(df.head())
```

```python
# Dataset information
print(df.info())
```

```python
# Statistical summary
print(df.describe())
```

```python
# Scatter plot
sns.scatterplot(x='age', y='expenses', data=df)
plt.show()
```

```python
# Line plot
sns.lineplot(x='age', y='expenses', data=df)
plt.show()
```

```python
# Histogram
sns.histplot(df['expenses'], kde=True)
plt.show()
```

```python
# Box plot
sns.boxplot(x=df['expenses'])
plt.show()
```

```python
# Correlation
print(df.corr(numeric_only=True))
```

```python
# Heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

```python
# Outlier visualization
sns.boxplot(y=df['expenses'])
plt.show()
```
