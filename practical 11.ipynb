# Clustering Fundamentals and Partitioning Methods: Cluster visualization, K-Medoids clustering implementation and cluster interpretation

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn_extra.cluster import KMedoids
```

```python
# Load dataset
df = pd.read_csv('Mall_Customers.csv')
```

```python
# Select features
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
# Create model
model = KMedoids(n_clusters=5, random_state=42)
```

```python
# Train model
model.fit(X)
```

```python
# Get clusters
df['Cluster'] = model.labels_
print(df.head())
```

```python
# Visualize clusters
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=model.labels_)
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('K-Medoids Clustering')
plt.show()
```

## second alternative code

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Load dataset
df = pd.read_csv('Mall_Customers.csv')
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
# Create model
model = KMeans(n_clusters=5, random_state=42, n_init=10)
```

```python
# Train model
model.fit(X)
```

```python
# Display clusters
df['Cluster'] = model.labels_
print(df.head())
```

```python
# Plot
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=model.labels_)
plt.show()
```
